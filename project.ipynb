```python
# Fill missing values with the median of each column
df_filled = df.fillna(df.median(numeric_only=True))

# Check that missing values are filled
print(df_filled.isnull().sum())
```

```python
# Summary statistics
print(df_filled.describe())
```

```python
# Class distribution
print(df_filled['Potability'].value_counts())
```

```python
sns.countplot(x='Potability', data=df_filled)
plt.title('Class Distribution of Potability')
plt.show()
```

```python
df_filled.hist(bins=30, figsize=(15, 10), edgecolor='black')
plt.suptitle('Feature Distributions', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
for col in df_filled.columns[:-1]:  # Skip 'Potability' #name boxplot
    plt.figure(figsize=(6, 4))
    sns.boxplot(x='Potability', y=col, data=df_filled)
    plt.title(f'{col} by Potability')
    plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df_filled.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
